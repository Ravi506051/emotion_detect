```python
import os
import zipfile
import glob

zip_files = glob.glob("/content/*.zip")

if not zip_files:
    print("ZIP file /content mein nahi mili. Step 1 ka output check karo.")
else:
    zip_path = zip_files[0]
    extract_path = "/content/face_detection_dataset"

    os.makedirs(extract_path, exist_ok=True)

    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall(extract_path)

    print("Dataset extracted successfully!")
    print("Extracted folder:", extract_path)
    print("Files:", os.listdir(extract_path)[:20])
```

```python
import os

image_extensions = (".jpg", ".jpeg", ".png", ".webp", ".bmp")

image_paths = []

for root, dirs, files in os.walk("/content/face_detection_dataset"):
    for file in files:
        if file.lower().endswith(image_extensions):
            image_paths.append(os.path.join(root, file))

print("Total images found:", len(image_paths))
print("First 10 image paths:")
for path in image_paths[:10]:
    print(path)
```

```python
import cv2
import matplotlib.pyplot as plt
from deepface import DeepFace

if not image_paths:
    print("Bhai, dataset mein images nahi mil rahi hain.")
else:
    image_path = image_paths[0]
    print("Processing:", image_path)

    try:
        result = DeepFace.analyze(
            img_path=image_path,
            actions=["emotion"],
            detector_backend="opencv",
            enforce_detection=False,
            silent=True
        )[0]

        image = cv2.imread(image_path)

        if image is None:
            print("Image open nahi ho paayi.")
        else:
            emotion = result["dominant_emotion"]
            score = result["emotion"][emotion]

            face = result["region"]
            x, y, w, h = face["x"], face["y"], face["w"], face["h"]

            if w > 0 and h > 0:
                cv2.rectangle(
                    image, (x, y), (x+w, y+h), (0, 255, 0), 2
                )

            cv2.putText(
                image, f"{emotion.upper()} ({score:.1f}%)",
                (max(x, 0), max(y-10, 25)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2
            )

            plt.figure(figsize=(8, 6))
            plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
            plt.axis("off")
            plt.show()

            print("Detected expression:", emotion)
            print("Model score:", round(score, 2), "%")

    except Exception as e:
        print("Error:", str(e))
```

```python
import os
import cv2
import pandas as pd
import zipfile
from deepface import DeepFace
from tqdm.auto import tqdm

# Output folders
output_dir = "/content/emotion_results"
os.makedirs(output_dir, exist_ok=True)

results = []

for i, image_path in enumerate(tqdm(image_paths)):
    try:
        predictions = DeepFace.analyze(
            img_path=image_path,
            actions=["emotion"],
            detector_backend="opencv",
            enforce_detection=False,
            silent=True
        )

        image = cv2.imread(image_path)

        if image is None:
            continue

        image_results = []

        for face_no, face_result in enumerate(predictions):
            emotion = face_result["dominant_emotion"]
            score = face_result["emotion"][emotion]
            region = face_result["region"]

            x = region["x"]
            y = region["y"]
            w = region["w"]
            h = region["h"]

            # Draw face box
            if w > 0 and h > 0:
                cv2.rectangle(
                    image, (x, y), (x + w, y + h),
                    (0, 255, 0), 2
                )

                cv2.putText(
                    image,
                    f"{emotion.upper()} ({score:.1f}%)",
                    (max(x, 0), max(y - 10, 20)),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6, (0, 255, 0), 2
                )

            image_results.append(emotion)

            results.append({
                "image_name": os.path.basename(image_path),
                "image_path": image_path,
                "face_number": face_no + 1,
                "predicted_emotion": emotion,
                "model_score_percent": round(float(score), 2)
            })

        # Save annotated image
        filename = f"result_{i:05d}.jpg"
        cv2.imwrite(os.path.join(output_dir, filename), image)

    except Exception as e:
        print(f"Skipped {image_path}: {e}")

# Save CSV
csv_path = "/content/emotion_results.csv"
df = pd.DataFrame(results)
df.to_csv(csv_path, index=False)

print("\nProcessing completed!")
print("Images found:", len(image_paths))
print("Faces processed:", len(df))
print("CSV saved at:", csv_path)

display(df.head(10))
```
